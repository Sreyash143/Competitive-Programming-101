## Set

There are 4 points we need to remember
1. Set is collection of unique items 
2. Set has no order items stored by hasing so no indexing / slicing allowed
2. set does not allow mutable data types inside it 
3. But set itself is mutable

### 1. create

```python
# set is a mathematic concept where a set we indicate a collection of unique items
S = {1,2,3,4,5}
print(type(S))

#fun fact we can create emplty set just like other containers like list and tyuple as in python by default there is dictionary it can be done only by type conversion 
S1 = {}
print(type(S1))

a = []
S2 = set(a)
print(S2 , type(S2))

#not only the duplicates are removed the order it also gone because all the elements are stored using hashing technique
S1 = {11,11,1,1,2,2,2,3,4,4,5,6,6} 
print(S1)
```

### 2. access (there is no way we can access any item we can only check)

```python
#we can't access any item only we can check if it's there in O(1)
if 1 in S1:
    print("yes")
else:
    print("no")
```

### 3. add

```python
# set does not allow it's items to be mutable because it will chnage their hashvalue and then finding them is not possible

# may be we can do it
S1.add(100) # adding one element 
print(S1)

# set does not allow mutable data types we tried to add list and set but fun part all of their elements are just added not list or set was added we can'add list

S1.update([6,3,4,8,9,0],{2,11,33,22}) # we can add list or multiple elements
print(S1)
```

```python
S1.add([2,3,4]) #unhashable type: 'list' it's not hashable 
```

### 4. delete

```python
S2 = {4,6,7,8}
print(S2)

S2.remove(4)# it will remove a particular item
print(S2)

print(S2.pop())#pop the last element but again last element as per hashing
print(S2)

S2.clear() # it will empty the set
print(S2)

del S2 # delele the set completely
print(S2)
```

### 5. set specific operations

```python
S3 = {1,2,3,4}
S4 = {3,4,5,6}

print(S3.union(S4))
print(S3.intersection(S4))
print(S3.symmetric_difference(S4))
print(S3.difference(S4))
print("\n")
print(S3)
print(S4)
```

```python
S3 = {1,2,3,4}
S4 = {3,4,5,6}
S3.difference_update(S4)
print(S3)

S3 = {1,2,3,4}
S4 = {3,4,5,6}
S3.intersection_update(S4)
print(S3)

S3 = {1,2,3,4}
S4 = {3,4,5,6}
S3.symmetric_difference_update(S4)
print(S3)
```

### 6. functions

```python
print(len(S1))
print(max(S3))
print(min(S3))
#set does not have count cause every element is guranteed to be unique
```

```python
print(S1.isdisjoint(S3))
print(S1.issubset(S3))
print(S1.issuperset(S3))
```

## Dictionary

There are some rules in dictionary
1. dictonary follows hashing technique so we can's really do indexing
2. for "keys" we can't put meutable elements it has to be mutable. best part we can actually acess element by their key value 
3. keys will be unique and immutable but values can be mutable (also keys can be different data tyep unlike other programming language)

### 1. create

```python
D = {"name":"Sreyash","age":21,"gender":"male","handsome":True}
print(D)

D1 = {"name":"bobo","age":17}
print(D1)
```

### 2. access

```python
print(D["name"])

# we can see the keys are only there in dictionary actually what happening is it's getting hashed so actually keys are there not their values
if "Sreyash" in D:
    print(True)
else:
    print(False)

if "age" in D:
    print(True)
else:
    print(False)
```

```python
#this is how we access step by step 2d and 3d dictionaries
D2 = {"name":"babi","age":21,"place":"kolkata","marks":{"eng":90,"math":100,"history":87}}
print(D2["name"])
print(D2["marks"])
print(D2["marks"]["history"])
D2.get("marks").get("eng")
```
