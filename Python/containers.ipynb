## List
There are steps we will be maintaining for each one of the container

### 1. create

```python
# List is can be heterogeneous 
# lists are mutable 
# unlike arrays they are not sitting in a continuous memory chunk in memeory
L = [1, "hello",[1,2,3,4],True]
print(L)
```

```python
L1 =[] #this is a empty list 
L2 = list("kolkata") # we can convert the type of other sequential data type
print(L1)
print(L2)
```

### 2. access

```python
print(L[1])
print(L[2][2]) # basically we can access the specific element by their index inside a list and for a list inside a list it's just doing indexing again
```


### 3. edit

```python
# as we know list are mutable so we actually can edit an list 
print(L)
L[1] = "Sreyash"
print(L)
```

### 4. add 

```python
# this is append (it allows you to add one element at the end of the list)
L.append("Sreyash")
print(L)
```

```python
# this is extend(it allows you to add multiple elements at the end of the list)
L.extend("kolkata")
print(L)
```

```python
# this is insert( it will allow you to insert any element in any position of the list best part rest of the elememts will move one index right)
L.insert(1,"babi")
print(L)
```

### 5. delete

```python
# this is del it means it will completely delete the list
del L1
```

```python
L1 # it should show error because I have deleted the list 
```

```python
#now it deleteing a exact position element from the list 
del L[-1] # it removed the last index from the list
print(L)
```

```python
# now it's remove means if we don't know the position but wanna to remove the element
print(L)
L.remove('Sreyash') # removed the first occurance of the element in case of multiple occurance
print(L)
```

```python
# pop means it will pop out the last element and give it to us 
print(L)
print(L.pop()) # removed the first occurance of the element in case of multiple occurance
print(L)
```

```python
#clear will clear the entire list and return an empty list
print(L2)
L2.clear()
print(L2)
```

### 6. operations

```python
# "+" allows two list to be concatinate and get stored in a new list
L1 = [1,3,4,"Sreyash"]
L2 = ["los angels"]
print(L1)
print(L2)
print(L1+L2)
```

```python
# "*" will help list to multiply
print(L1*3) # it's a new list original is still intact
print(L1)
```

```python
# membarship operator works fine 
for i in L1:
    print(i*2)# we can even 
```

### 7. functions

```python
list1 = [1,2,3,4]
list2 = [1,2,3,4]
print(id(list1),id(list2)) #means these two are located in different memeory address unlike string or normal variables
```

```python
#copy will help in copy every element from a list to a new list
L3 = L.copy()
print(L3)
print(L == L3)#this thing compares the value
print(L is L3) # this is the proof that two list are separate though I have just copied the value these are two different location in memory
```

```python
list1 = [3,19,4,2,8]
print(sorted(list1,reverse= True))# this is returning sorted list but it's a new list not updateing original list 
print(list1)
list1.sort()# this is returning none but permanenly updating the original list 
print(list1)
list1.reverse()
print(list1)
```
