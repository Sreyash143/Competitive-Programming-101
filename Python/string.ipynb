### Playing with characters case in the String 

One thing to always remember ***Strings*** are immutable so everytime we prform some operation it will create a new string it is not modifing the old string

```python
a = "I am a boy."
a.capitalize()
a.upper()
a.lower()
a.swapcase()
a.casefold()
```

### Find and Index fucntion 

This function is very important ion finding any character or charaters in a **string** for multiple occurance it will just provide first occurance
1. find() this will try to find if could not find return -1 also
2. index() this will try to find and return index but if not then throw an error

```python
a = "Sreyash"
a.find('p')
a.index('p')
```

### starts with and ends with these two are to find out any string is starting or ending with some character/characters

```python
a.startswith("Sre")
a.endswith("sh")
```

### count the number of particular character in the string

```python
a.count('r')
```
