### memory for different data types

```python
import sys

#integer
x = 0
a = 7
b = 10**100000
print(f"memeory size for {x} is {sys.getsizeof(x)} bytes")
print(f"memeory size for {a} is {sys.getsizeof(a)} bytes")
print(f"memeory size for b is {sys.getsizeof(b)} bytes") # now unlike other languages we can see for integer for enormous values the byte size is increasing using arbitary precision arithmatic 
print()

#float (approx 24 bytes)
c = 8.9
d = 6.89765
print(f"memeory size for {c} is {sys.getsizeof(c)} bytes")
print(f"memeory size for {d}is {sys.getsizeof(d)} bytes")

print()
# boolean
e = True
f = False
print(f"memeory size for {e} is {sys.getsizeof(e)} bytes")
print(f"memeory size for {f} is {sys.getsizeof(f)} bytes")

print()

#none takes 16 bytes
print(f"memeory size for None is {sys.getsizeof(None)} bytes")
```

```python
me =""
name = "Sreyash"
description = "Hello_I_am_a_boy"
print(f"memeory size for empty string is {sys.getsizeof(me)} bytes")# an empty string always takes 49 bytes
print(f"memeory size for name is {sys.getsizeof(name)} bytes")# everytime we ad something it will add that many bytes as many chacters are there in the string 
print(f"memeory size for description is {sys.getsizeof(description)} bytes")
```

```python
# starting from 56 bytes everytime you add element it will grow by 8 bytes
L =[]
print(f"memory size of empty list is {sys.getsizeof(L)}")
for i in range(10):
    L.append(i)
    print(f"Items: {len(L)} | Memory size: {sys.getsizeof(L)} bytes")


# best part is it's only storing the address of the element in a continuous memeory block that's why it can allocate different datatype unlike other programming language datastructure
```

### Referencing 

```python
# everytime we create any variable we are actually pointing a pointer to the memory adreess the exact element is present now we need to understand if we just change any variable the value is not changing now the variable is actually pointing to a different address 


a = 6
b = a
c = b
print(id(a))
print(id(b))
print(id(c))
#  all of these addreses are same this is called aliasing
print()

a = a+1
print(id(a))
print(id(b))
print(id(c))
```

```python
# -5 to 256 this range values are most frquently used so python already stores these values in memory from the beginning this is called interning here both of the variables are poiting to the same address
a = 256
b = 256

print(id(a))
print(id(b))

print()

#that's why here both of the variable are not pointing to the same address
a = 257
b = 257
print(id(a))
print(id(b))
```

```python
# how many variables we have created and getrefcount so it should give us varaible number + 1 
x = 50705
y = x
z = y
print(sys.getrefcount(x))

a = 3
b = 103

print(sys.getrefcount(a))
print(sys.getrefcount(b))# actually many other programs are using this number that's why total number of variblas pointing to this number is this much high
```

```python
# only those strings which is valid identifier for them variables are pointng to the same address but if not valid identifier then they are not pointing to same address
a = "hello"
b = "hello"
print(id(a))
print(id(b))


print()

a = "I am sreyash"
b = "I am Sreyash"
print(id(a))
print(id(b))
```
